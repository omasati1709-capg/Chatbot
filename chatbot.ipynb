```python
import os
from datetime import datetime

from openai import OpenAI
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

# Initialize OpenAI client
client = OpenAI(
    base_url="https://openai.generative.engine.capgemini.com/v1",
    api_key=os.getenv("AI_KEY")
)

HISTORY_FILE = "history.txt"


# Load previous chat history
def load_history():
    if os.path.exists(HISTORY_FILE):
        with open(HISTORY_FILE, "r", encoding="utf-8") as file:
            return file.read()
    return ""


# Save chat history
def save_history(user_message, bot_response):
    with open(HISTORY_FILE, "a", encoding="utf-8") as file:
        file.write(f"\nUser: {user_message}\n")
        file.write(f"Bot: {bot_response}\n")


# Create a new txt file for each conversation
def save_conversation_to_file(user_message, bot_response):

    timestamp = datetime.now()

    safe_question = "".join(
        c for c in user_message[:30]
        if c.isalnum() or c in (" ", "_")
    )

    safe_question = safe_question.strip().replace(" ", "_")

    filename = (
        f"{safe_question}_"
        f"{timestamp.strftime('%Y%m%d_%H%M%S')}.txt"
    )

    with open(filename, "w", encoding="utf-8") as file:
        file.write(
            f"Timestamp: "
            f"{timestamp.strftime('%Y-%m-%d %H:%M:%S')}\n\n"
        )

        file.write("USER QUESTION\n")
        file.write("-" * 50 + "\n")
        file.write(user_message + "\n\n")

        file.write("BOT ANSWER\n")
        file.write("-" * 50 + "\n")
        file.write(bot_response)

    return filename


# Generate chatbot response
def chatbot(user_message, history):

    response = client.chat.completions.create(
        model="qwen.qwen3-coder-next",
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a helpful chatbot. "
                    "Use previous conversation context when answering."
                )
            },
            {
                "role": "user",
                "content": f"""
Previous conversation:
{history}

Current question:
{user_message}
"""
            }
        ]
    )

    return response.choices[0].message.content


# Main program
print("Chatbot started! Type 'exit' to stop.")

while True:

    user_input = input("\nYou: ")

    if user_input.lower() == "exit":
        print("\nBot: Goodbye!")
        break

    history = load_history()

    try:

        bot_response = chatbot(user_input, history)

        print("\nBot:", bot_response)

        # Save complete history
        save_history(user_input, bot_response)

        # Create separate txt file
        file_name = save_conversation_to_file(
            user_input,
            bot_response
        )

        print(f"\nSaved to file: {file_name}")

    except Exception as e:
        print("\nError:", e)
```

```python
import os
from datetime import datetime

from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

client = OpenAI(
    base_url="https://openai.generative.engine.capgemini.com/v1",
    api_key=os.getenv("AI_KEY")
)

CONVERSATION_FOLDER = "conversations"

# Create folder if not exists
os.makedirs(CONVERSATION_FOLDER, exist_ok=True)


def load_memory():
    """
    Load all previous conversations.
    """

    memory = ""

    files = sorted(os.listdir(CONVERSATION_FOLDER))

    for file in files:
        if file.endswith(".txt"):
            with open(
                os.path.join(CONVERSATION_FOLDER, file),
                "r",
                encoding="utf-8"
            ) as f:
                memory += f.read() + "\n\n"

    return memory


def save_conversation(user_message, bot_response):

    timestamp = datetime.now()

    filename = (
        f"chat_{timestamp.strftime('%Y%m%d_%H%M%S')}.txt"
    )

    filepath = os.path.join(
        CONVERSATION_FOLDER,
        filename
    )

    with open(filepath, "w", encoding="utf-8") as f:

        f.write(
            f"Timestamp: "
            f"{timestamp.strftime('%Y-%m-%d %H:%M:%S')}\n\n"
        )

        f.write(
            f"User: {user_message}\n\n"
        )

        f.write(
            f"Bot: {bot_response}\n"
        )

    return filename


def chatbot(user_message):

    memory = load_memory()

    response = client.chat.completions.create(
        model="qwen.qwen3-coder-next",
        messages=[
            {
                "role": "system",
                "content": """
You are a personal assistant chatbot.

Use previous conversations as memory.

If the user asks:
- What did I ask earlier?
- What projects am I working on?
- Summarize our discussion

Use the provided memory.
"""
            },
            {
                "role": "user",
                "content": f"""
Previous Memory:

{memory}

Current User Question:

{user_message}
"""
            }
        ]
    )

    return response.choices[0].message.content


print("Chatbot started! Type 'exit' to stop.")

while True:

    user_input = input("\nYou: ")

    if user_input.lower() == "exit":
        print("\nBot: Goodbye!")
        break

    try:

        bot_response = chatbot(user_input)

        print(f"\nBot: {bot_response}")

        filename = save_conversation(
            user_input,
            bot_response
        )

        print(
            f"\nSaved conversation: {filename}"
        )

    except Exception as e:
        print("\nError:", e)
```

```python
import os
from datetime import datetime
from dotenv import load_dotenv
from openai import OpenAI

# Load environment variables
load_dotenv()

# Initialize client
client = OpenAI(
    base_url="https://openai.generative.engine.capgemini.com/v1",
    api_key=os.getenv("AI_KEY")
)

# Folder containing all conversations
CONVERSATION_FOLDER = "conversations"

# Create folder if it doesn't exist
os.makedirs(CONVERSATION_FOLDER, exist_ok=True)


def load_memory():
    """
    Reads all previous conversation files and returns them as memory.
    """

    memory = ""

    files = sorted(os.listdir(CONVERSATION_FOLDER))

    for file in files:
        if file.endswith(".txt"):

            filepath = os.path.join(
                CONVERSATION_FOLDER,
                file
            )

            with open(
                filepath,
                "r",
                encoding="utf-8"
            ) as f:
                memory += f.read()
                memory += "\n\n"

    return memory


def save_conversation(user_question, bot_answer):
    """
    Save current conversation into a new txt file.
    """

    timestamp = datetime.now()

    filename = (
        f"chat_{timestamp.strftime('%Y%m%d_%H%M%S')}.txt"
    )

    filepath = os.path.join(
        CONVERSATION_FOLDER,
        filename
    )

    with open(
        filepath,
        "w",
        encoding="utf-8"
    ) as file:

        file.write(
            f"Timestamp: "
            f"{timestamp.strftime('%Y-%m-%d %H:%M:%S')}\n\n"
        )

        file.write("USER QUESTION\n")
        file.write("=" * 50 + "\n")
        file.write(user_question)

        file.write("\n\n")

        file.write("BOT ANSWER\n")
        file.write("=" * 50 + "\n")
        file.write(bot_answer)

    return filename


def chatbot(user_question):

    memory = load_memory()

    response = client.chat.completions.create(
        model="qwen.qwen3-coder-next",
        messages=[
            {
                "role": "system",
                "content": """
You are a helpful personal chatbot.

You have access to previous conversations.

When relevant, use previous conversations
to answer questions and remember facts
previously shared by the user.
"""
            },
            {
                "role": "user",
                "content": f"""
Previous Conversations:

{memory}

Current User Question:

{user_question}
"""
            }
        ]
    )

    return response.choices[0].message.content


def main():

    print("=" * 60)
    print("MEMORY CHATBOT")
    print("=" * 60)

    user_question = input("\nYou: ")

    if not user_question.strip():
        print("Question cannot be empty.")
        return

    try:

        bot_answer = chatbot(user_question)

        print("\nBot:")
        print(bot_answer)

        filename = save_conversation(
            user_question,
            bot_answer
        )

        print(
            f"\nConversation saved successfully:"
        )
        print(
            os.path.join(
                CONVERSATION_FOLDER,
                filename
            )
        )

    except Exception as e:
        print("\nError:", str(e))


if __name__ == "__main__":
    main()
```
